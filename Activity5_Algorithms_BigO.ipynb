```python
def linear_operations(n):
    count = 0
    for i in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n^2)={n*n:7d}"
    )
```

```python
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, 
    (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100, 
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)
print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10
performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time)
]

print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>12}")
for name, result, elapsed in results:
    print(
        f"{name:<18}{result[1]:>12}"
        f"{result[2]:>12}{elapsed:>12.8f}"
    )
```

```python
# Student-Specific Inputs
LAST_NAME = "Abalos"
STUDENT_ID = "TUPM-26-1873"
FAVORITE_ARTIST = "Sabrina Carpenter"

# Rule Calculations
SEED_DIGIT = int(STUDENT_ID[-1])                   # Last digit of student ID (3)
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # Sum of ID digits (27)
NAME_LENGTH = len(LAST_NAME)                       # Length of surname (6)
ARTIST_LENGTH = len(FAVORITE_ARTIST)               # Length of artist name (17)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha()) # Alphabetical sum (A=1...Z=26)

print(f"Seed Digit: {SEED_DIGIT}")
print(f"ID Sum: {ID_SUM}")
print(f"Name Length: {NAME_LENGTH}")
print(f"Artist Length: {ARTIST_LENGTH}")
print(f"Artist Sum: {ARTIST_SUM}")
```

```python
# 1. Generate original and rotated datasets
original_dataset = sorted([(ID_SUM * i + SEED_DIGIT) % 100 for i in range(1, 11)])
rotation_point = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_point:] + original_dataset[:rotation_point]
target_incident = (ID_SUM + ARTIST_SUM) % 100

# 2. Search for target using linear search and track comparisons
def search_rotated(data, target):
    comparisons = 0
    for i, val in enumerate(data):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

index_pos, comp_count = search_rotated(rotated_dataset, target_incident)

print("--- EXERCISE 1 RESULTS ---")
print(f"Original Dataset: {original_dataset}")
print(f"Rotated Dataset: {rotated_dataset}")
print(f"Target Incident: {target_incident}")
print(f"Position Index: {index_pos if index_pos != -1 else 'Not Found'}")
print(f"Comparison Count: {comp_count}")
print("Time Complexity: O(n) | Space Complexity: O(1)")
```

```python
# 1. Generate two sorted enrollment datasets
dataset_a = sorted([(SEED_DIGIT * i + ID_SUM) % 100 for i in range(1, 6)])
dataset_b = sorted([(ARTIST_LENGTH * i + NAME_LENGTH) % 100 for i in range(1, 6)])

# 2. Consolidate into one ordered collection (Merge algorithm)
def merge_datasets(a, b):
    merged = []
    i = j = 0
    comparisons = 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    while i < len(a):
        merged.append(a[i])
        i += 1
    while j < len(b):
        merged.append(b[j])
        j += 1
    return merged, comparisons

consolidated_dataset, merge_comparisons = merge_datasets(dataset_a, dataset_b)

print("--- EXERCISE 2 RESULTS ---")
print(f"Dataset A: {dataset_a}")
print(f"Dataset B: {dataset_b}")
print(f"Consolidated Dataset: {consolidated_dataset}")
print(f"Records Before: {len(dataset_a) + len(dataset_b)} | Records After: {len(consolidated_dataset)}")
print(f"Comparison Count: {merge_comparisons}")
print("Time Complexity: O(n + m) | Space Complexity: O(n + m)")
```

```python
import time

base_orders = [(ID_SUM * 2 + i * 3) % 100 for i in range(12)]
inputs = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}

def bubble_sort(data):
    a, comps, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comps, swaps

def selection_sort(data):
    a, comps, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        min_idx = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            swaps += 1
    return a, comps, swaps

def insertion_sort(data):
    a, comps, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comps, shifts

print("--- EXERCISE 3 RESULTS ---")
for cond_name, data_val in inputs.items():
    print(f"\nCondition: {cond_name} (Data: {data_val})")
    b_res, b_comps, b_swaps = bubble_sort(data_val)
    s_res, s_comps, s_swaps = selection_sort(data_val)
    i_res, i_comps, i_shifts = insertion_sort(data_val)
    
    print(f"  Bubble Sort    -> Comparisons: {b_comps:2d}, Swaps: {b_swaps:2d}")
    print(f"  Selection Sort -> Comparisons: {s_comps:2d}, Swaps: {s_swaps:2d}")
    print(f"  Insertion Sort -> Comparisons: {i_comps:2d}, Shifts: {i_shifts:2d}")
```
